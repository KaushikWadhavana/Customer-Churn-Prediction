```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/Customer Churn.csv")
```

```python
df.head()
```

```python
missing_values = df.isnull().sum()
```

```python
print("Blank TotalCharges:", (df["TotalCharges"].astype(str).str.strip() == "").sum())
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
print("Missing TotalCharges After Conversion:",
      df["TotalCharges"].isnull().sum())
```

```python
print(df.loc[
    df["TotalCharges"].isnull(),
    ["customerID", "tenure", "TotalCharges"]
])
```

```python
mask = df["TotalCharges"].isnull() & (df["tenure"] == 0)
df.loc[mask, "TotalCharges"] = 0
```

```python
print("Remaining Missing TotalCharges:",
      df["TotalCharges"].isnull().sum())
```

```python
print("Duplicates Before Cleaning:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates After Cleaning:", df.duplicated().sum())
```

```python
print("Final Dataset Shape:", df.shape)
print("\nRemaining Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df.to_csv("../data/cleaned/customer_churn_cleaned.csv", index=False)
```
