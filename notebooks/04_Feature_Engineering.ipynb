```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/cleaned/customer_churn_cleaned.csv")
print("Dataset Shape:", df.shape)
df.head()
```

```python
df_fe = df.copy()

print("Feature Engineering Dataset Shape:", df_fe.shape)
```

```python
df_fe = df_fe.drop(columns=["customerID"])
```

```python
df_fe.head()
```

```python
df_fe["Churn"] = df_fe["Churn"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df_fe["Churn"].value_counts()
```

```python
## Identify numerical and categorical features
```

```python
numerical_columns = df_fe.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()
```

```python
categorical_columns = df_fe.select_dtypes(
    include=["object"]
).columns.tolist()
```

```python
# Churn is the target, not an input feature
```

```python
if "Churn" in numerical_columns:
    numerical_columns.remove("Churn")

print("Numerical Features:")
print(numerical_columns)

print("\nCategorical Features:")
print(categorical_columns)
```

```python
## Encode categorical features
```

```python
# Separate input features and target
```

```python
X = df_fe.drop(columns=["Churn"])
y = df_fe["Churn"]
```

```python
# Convert categorical columns into numerical columns
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)

print("Encoded Feature Shape:", X_encoded.shape)
X_encoded.head()
```

```python
df_features = X_encoded.copy()
df_features["Churn"] = y
```

```python
df_features.to_csv(
    "../data/cleaned/customer_churn_features.csv",
    index=False
)
```
